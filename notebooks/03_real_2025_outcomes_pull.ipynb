```python
draft_board = season_totals.filter(pl.col("season") == 2025).select([
    "player_display_name", "position", "team", "games_played",
    "ppr_per_game", "ppr_std_dev", "target_share_pct",
    "yards_per_carry", "yards_per_reception"
])

draft_board.head()
```

```python
draft_board = draft_board.join(
    established_trends.select(["player_display_name", "ppr_change_24_to_25"]),
    on="player_display_name",
    how="left"
)

draft_board.head()
```

```python
draft_board = draft_board.with_columns([
    pl.col("ppr_per_game").rank(method="ordinal", descending=True).over("position").alias("position_rank")
])

draft_board.sort(["position", "position_rank"]).head(15)
```

```python
draft_board = draft_board.with_columns([
    pl.when(pl.col("position_rank") <= 3).then(1)
    .when(pl.col("position_rank") <= 8).then(2)
    .when(pl.col("position_rank") <= 15).then(3)
    .when(pl.col("position_rank") <= 24).then(4)
    .otherwise(5)
    .alias("tier")
])

draft_board.sort(["position", "position_rank"]).head(15)
```

```python
draft_board = draft_board.with_columns([
    pl.col("ppr_change_24_to_25").fill_null(0).alias("ppr_change_filled")
])

draft_board = draft_board.with_columns([
    (
        pl.col("ppr_per_game")
        - (0.3 * pl.col("ppr_std_dev").fill_null(0))
        + (0.3 * pl.col("ppr_change_filled"))
    ).alias("composite_score")
])

draft_board.select(["player_display_name", "position", "ppr_per_game", "ppr_std_dev", "ppr_change_24_to_25", "composite_score"]).sort("composite_score", descending=True).head(10)
```

```python
draft_board = draft_board.with_columns([
    pl.col("composite_score").rank(method="ordinal", descending=True).over("position").alias("position_rank")
])

draft_board = draft_board.with_columns([
    pl.when(pl.col("position_rank") <= 3).then(1)
    .when(pl.col("position_rank") <= 8).then(2)
    .when(pl.col("position_rank") <= 15).then(3)
    .when(pl.col("position_rank") <= 24).then(4)
    .otherwise(5)
    .alias("tier")
])

draft_board.sort(["position", "position_rank"]).head(15)
```

```python
replacement_levels = {}

for pos, cutoff in [("QB", 20), ("RB", 31), ("WR", 31), ("TE", 14)]:
    player_at_cutoff = draft_board.filter(
        (pl.col("position") == pos) & (pl.col("position_rank") == cutoff)
    )
    replacement_levels[pos] = player_at_cutoff["composite_score"][0]

replacement_levels
```

```python
draft_board = draft_board.with_columns([
    pl.when(pl.col("position") == "QB").then(pl.col("composite_score") - replacement_levels["QB"])
    .when(pl.col("position") == "RB").then(pl.col("composite_score") - replacement_levels["RB"])
    .when(pl.col("position") == "WR").then(pl.col("composite_score") - replacement_levels["WR"])
    .when(pl.col("position") == "TE").then(pl.col("composite_score") - replacement_levels["TE"])
    .alias("vorp")
])

draft_board.sort("vorp", descending=True).select(
    ["player_display_name", "position", "composite_score", "vorp"]
).head(15)
```
